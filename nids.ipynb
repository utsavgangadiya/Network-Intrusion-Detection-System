Exploratory Data Analysis (EDA)

```python
plt.figure(figsize=(10,6))

sns.countplot(
    y=df["Attack Type"],
    order=df["Attack Type"].value_counts().index
)

plt.title("Attack Type Distribution")
plt.tight_layout()

plt.savefig("charts/attack_distribution.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(8,8))

df["Attack Type"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Attack Type Percentage")

plt.savefig("charts/attack_pie_chart.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(20,18))

corr = df.select_dtypes(include=np.number).corr()

sns.heatmap(
    corr,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.savefig("charts/correlation_heatmap.png", dpi=300)

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(df["Flow Duration"], bins=30)

plt.title("Flow Duration Distribution")

plt.savefig("charts/flow_duration_histogram.png", dpi=300)

plt.show()
```

```python
corr = df.select_dtypes(include=np.number).corr()

target_corr = corr["Packet Length Mean"].sort_values(ascending=False)

plt.figure(figsize=(8,10))

target_corr.head(15).plot(kind="barh")

plt.title("Top Correlated Features")

plt.tight_layout()

plt.savefig("charts/top_correlations.png", dpi=300)

plt.show()
```

```python
summary = pd.DataFrame({
    "Count": df["Attack Type"].value_counts(),
    "Percentage": round(df["Attack Type"].value_counts(normalize=True)*100,2)
})

print(summary)

summary.to_csv("charts/attack_summary.csv")
```
